3. Create a function taking one parameters for a password. The function would convert each character of the password into corresponding ASCII numbers and return one string composed of the ASCII numbers instead

````python
pw=input('Enter a password:')
def asciipass(pw):
    acpass=''
    for i in pw:
        b=str(ord(i))
        acpass+=b
    return acpass
print(asciipass(pw))
    
````

4. Create a function (or set of functions ) taking one parameter for a password.

  - the password contains at least one lower case letter,

````python
pw=input('Enter a password:')
def password(pw):
    if " " in pw:
        print('Password should not contain spaces.')
    upper=0
    lower=0
    digit=0
    special=0
    for i in pw:
        if i.isupper():
            upper=1
        elif i.islower():
            lower=1
        elif i.isdigit():
            digit=1
        elif i in '!@#$%^&*()-_=+[]{}|:;?':
            special=1
    if upper==1 and lower==1 and digit==1 and special==1:
        return True
    else:
        return False
password(pw)
````

````python
pw='P@#yn26at^&i5ve'
def password(pw):
    upper=[]
    lower=[]
    digit=[]
    special=[]
    for i in pw:
        if i.isupper():
            upper.append(i)
        elif i.islower():
            lower.append(i)
        elif i.isdigit():
            digit.append(i)
        elif i in '!@#$%^&*()-_=+[]{}|:;?':
            special.append(i)
    print('The total count of Upper case is:',len(upper),'-',upper)
    print('The total count of Lower case is:',len(lower),'-',lower)
    print('The total count of Digits is:',len(digit),'-',digit)
    print('The total count of Special Characters is:',len(special),'-',special)
password(pw)
````

13. Secret Word Generator with Randomization: Ask the user for their first name and surname. Create a function that generates a special secret word that:
Uses the first 3 letters of the first name and last 2 letters of the surname.
Randomly appends a special character (e.g., @, #, $, etc.) at the end of the secret word.

```python
special_chars = "@#$%^&*!"
```

````python
import random as r
n=input('Enter your Name:')
sn=input('Enter your Surname:')
def secretword(n,sn):
    a=n[:3]
    b=sn[-2:]
    c=a+b
    special_chars = "@#$%^&*!"
    c+=r.choice(special_chars)
    return c
print('Your secret word is:',secretword(n,sn))
````
